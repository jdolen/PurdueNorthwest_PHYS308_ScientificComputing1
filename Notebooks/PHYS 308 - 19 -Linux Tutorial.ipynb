# Linux for Physics

## Introduction: Why Linux?

The vast majority of scientific computing—from the Large Hadron Collider (LHC) to astrophysical simulations and condensed matter modeling—runs on **Linux**. A few reasons why:


1.  **High Performance Computing (HPC):** Physicists often analyze huge datasets requiring heavy calculation.
This scale of computing must be performed with giant clusters of computers or supercomputers, which you almost exclusively access via a text-based Linux terminal.
2.  **Automation:** You can script a command to process thousands of files with a single line of code.
3.  **Reproducibility:** Text-based scripts ensure your analysis methodology is transparent and can be repeated exactly by other researchers.
4.  **The Scientific Software Ecosystem:** While commercial software often runs on Windows or Macs, the open-source "heavy lifting" tools in physics are native to Linux.
    * **Particle Physics:** The standard tools for analyzing collision data (**ROOT**) and simulating detector interactions (**Geant4**) were built for Linux.
    * **Astrophysics:** Radio astronomy suites (like **CASA**) and X-ray analysis tools (like **HEASoft**), as well as massive N-body cosmological simulations, run primarily on Linux environments.
    * **Condensed Matter:** If you study materials, you will likely use Molecular Dynamics or Density Functional Theory (DFT) codes like **LAMMPS**, **Quantum ESPRESSO**, or **VASP**. These are designed to run in parallel on Linux clusters to model atomic structures.
    * **Python & C++:** While cross-platform, these languages often perform better and are easier to manage (via package managers) in a Linux environment for high-performance tasks.
5.  **Distributed Computing & The Grid:** Modern physics relies on the **Open Science Grid (OSG)** and the **WLCG** (Worldwide LHC Computing Grid).
This is a global collaboration of more than 170 computing centers in 42 countries, linking up linux clusters at national laboratories and universities.
  


## History

### Unix

In 1969, Bell Labs researchers Ken Thompson and Dennis Ritchie created Unix.They established the "Unix Philosophy". That is, building small, modular tools that do one thing well (like "sort a file" or "find text") and chaining them together using pipes (|). This allowed physicists to build complex data analysis workflows without writing massive, monolithic programs

In 1973, Unix was rewritten in the C programming language. This was revolutionary. Before this, operating systems were hard-coded to specific hardware. Unix was "portable," meaning it could be installed on different machines.

Because AT&T (Bell Labs' parent) was a regulated monopoly, they couldn't sell software for profit. Instead, they licensed Unix to universities for a nominal fee. By the late 1970s, a generation of computer scientists and physicists were growing up on Unix.

### Unix and the internet

The marriage of Unix and the internet occurred in the early 1980s when the University of California, Berkeley, released a version of Unix (BSD Unix) with the TCP/IP networking protocols built directly into the core system. This was funded by DARPA specifically to help researchers connect to the ARPANET. Because the networking stack was free and integrated, Unix became the default operating system of the early internet. By the mid-1980s, if a university or National Lab wanted to connect to the NSFNET backbone to access supercomputing centers, they almost certainly used Unix-based workstations.

### The Web

The invention of the World Wide Web was a direct application of this Unix-based networking capability. When Tim Berners-Lee created the Web at CERN in 1989, he wrote the first browser and server software on a NeXT computer—a workstation running a Unix-derivative OS. The multi-tasking and networking capabilities inherent to Unix made it the only viable platform at the time for serving hypermedia to multiple users simultaneously.

U.S. National Labs were the first to import this technology because they were already immersed in the Unix ecosystem. The multi-tasking and networking capabilities of Unix allowed the first web server to handle multiple requests from different physicists simultaneously.

In 1991, SLAC physicist Paul Kunz visited CERN, saw the web, and immediately installed a server on SLAC’s IBM mainframe (running a Unix-like environment) to host the SPIRES database. This allowed physicists globally to search for preprints remotely. Fermilab followed suit in 1992, setting up a server to distribute data from the Tevatron.


### Linux

While Unix provided the architecture, it was expensive. Workstations from Sun or SGI could cost $20,000 to $50,000, severely limiting how many a physics department could buy. This created a massive hunger for a Unix-like system that could run on cheap, commodity PC hardware (Intel x86 chips).

Enter Linux. Created by Linus Torvalds in 1991, Linux was adopted by the scientific community almost instantly—far faster than in the corporate world.

Linux was adopted because it provided the power, scripting tools, and networking of Unix, but it was free and open-source. Scientists could modify the kernel to handle specific hardware or data acquisition tasks, which was impossible with proprietary Windows or Mac OS.

### Computing clusters

A computing cluster is essentially a group of separate computers—often called "nodes"—that are connected by a network and work together to act as a single, powerful system.

The defining feature of a cluster is that it is built using commodity hardware. This means the nodes are often standard, off-the-shelf computers—similar to the desktop PC you might have at home. This approach, pioneered by the "Beowulf" clusters at NASA in the 1990s, democratized high-performance computing because it allowed universities to build powerful systems cheaply by wiring together hundreds of mass-produced CPUs


By the late 1990s, Linux had become the standard OS for high-energy physics and astronomy. This culminated in "Scientific Linux," a distribution co-developed by Fermilab and CERN specifically to provide a stable, standard platform for analyzing the massive datasets produced by the Large Hadron Collider and the Tevatron.

Clusters are different that supercomputers. In the 1970s and 80s supercomputers were monolithic. They were custom-built from the ground up with proprietary processors and specialized circuitry designed to fit into a specific physical chassis. You couldn't just go to a store and buy a "part" for a Cray-1; every inch of it was engineered for pure speed. In modern clusters the difference between a cluster and supercomputer is a bit blurred, but generally standard clusters connect using standard Ethernet and are generally very useful when the computers don't need to speak to each other, while supercomputers use ultra-low-latency interconnects, which are great for simulations where the comptuters have to talk to each other (ex. gravitational physics simulations where the calculation of one computer depends on the resutls of another).

### Unix vs Linux

The experience of using a tradidional Unix machine is 95% identical to a Linux machine. Linux was explicitly designed to be "Unix-like."

## Part 6 -  Wildcards
You will often generate thousands of files named `run1.dat`, `run2.dat`, etc. Wildcards allow you to select specific groups of files.

First, let's create a messy environment to test this:
```bash
touch run1.dat run2.dat run10.dat run20.dat noise.txt test.log experiment_A.data experiment_B.data
```

### 1. The Asterisk (`*`)
Matches **zero or more** characters. This is the most common wildcard.
* `ls *.dat` : Lists all files ending in .dat
* `ls run*` : Lists all files starting with "run"

```bash
ls *.dat
```

### 2. The Question Mark (`?`)
Matches exactly **one** character.
* `ls run?.dat` : Will match `run1.dat` but **not** `run10.dat` (because 10 is two characters).

```bash
ls run?.dat
```

### 3. Brackets (`[]`)
Matches a **range** or specific set of characters.
* `ls run[1-3]*` : Matches files starting with run1, run2, or run3.

### 4. Braces (`{}`)
Used to expand sequences (Bash expansion). This is great for batch creation.
* `touch data_{1..5}.txt` : Creates data_1.txt through data_5.txt.
* `ls *.{dat,log}` : Lists files ending in EITHER .dat OR .log.

```bash
ls *.{dat,log}
```

## Part 7: Grep, Pipes, and Sed

### 1. `grep`

`grep` searches for specific text patterns within files.

**Setup:** Create a fake log file.
```bash
cat <<EOF > detector.log
INFO: Detector cooling system active
INFO: Voltage stable at 2000V
WARNING: Temperature fluctuation detected in Sector 7
INFO: Beam stable
ERROR: Sensor 4 failure
ERROR: HV trip
INFO: End of run
EOF
```

**Usage:**
* `grep "ERROR" detector.log` : Prints lines containing "ERROR".
* `grep -v "INFO" detector.log` : Invert match. Prints everything that is **NOT** "INFO".
* `grep -i "error" detector.log` : Case insensitive search.

```bash
grep "ERROR" detector.log
```

### 2. The Pipe (`|`)
The pipe takes the output of the command on the left and feeds it as input to the command on the right. This allows you to chain tools together.

*Example: List all files, find the ones containing "run", and count them (`wc -l`).*
```bash
ls | grep "run" | wc -l
```

### 3. `sed`: Editing streams
`sed` allows you to find and replace text programmatically.
*Syntax:* `s/find/replace/g` (g = global, do it for all instances in the line).

*Example: Your code expects units in MeV, but the file says GeV.*
```bash
sed 's/GeV/MeV/g' calibration_data.csv
```
*Note: To save this change, you must redirect it to a new file:*
```bash
sed 's/GeV/MeV/g' calibration_data.csv > calibration_fixed.csv
```

## Part 8: Shell Scripting
Never do manually what you can automate. A shell script is just a text file containing a list of commands.

**1. Create the script**
We will create a script that simulates analyzing 3 different data runs.

```bash
cat <<EOF > analysis_bot.sh
#!/bin/bash
# The first line (Shebang) tells the system to use bash to run this.

echo "Initiating analysis sequence..."

# Loop from 1 to 3
for i in 1 2 3
do
    echo "Processing Run \$i..."
    
    # Create a directory for this run
    mkdir -p "Data_Run_\$i"
    
    # Simulate generating a result file
    echo "Analysis complete for Run \$i" > "Data_Run_\$i/results.txt"
    
    # Add a timestamp
    date >> "Data_Run_\$i/results.txt"
done

echo "All runs processed."
EOF
```

**2. Permissions (`chmod`)**
By default, for security, scripts are not allowed to run. You must make them executable.
* `chmod` = Change Mode
* `+x` = Add eXecute permission

```bash
chmod +x analysis_bot.sh
```

**3. Run it**
```bash
./analysis_bot.sh
```


Now try creating and running this script.

```bash
cat <<EOF > batch_rename.sh
#!/bin/bash

for file in run*; do
    echo $file

    # 1. Echo the filename into sed
    # 2. Use 's' (substitute) to replace 'run' with 'spill'
    # 3. Save the result into a variable
    new_name=$(echo "$file" | sed 's/^run/spill/')
    
    # Check if the name actually changed (avoids errors)
    if [ "$file" != "$new_name" ]; then
        mv "$file" "$new_name"
    fi
done

EOF
```

## Part 10 -  Redirection: Saving Your Work

By default, every command you type sends its results to **Standard Output** (stdout), which is just a fancy name for "your screen."

But sometimes, you don't want to see the output right now. You want to save it to a file to use later. This is called **Redirection**.

### The Overwrite Operator (`>`)
The greater-than symbol `>` tells the computer: "Take the output of this command and dump it into a file."

**Warning:** This will **create** a new file or **overwrite** an existing one completely.

```bash
# Save the list of files to a text file
ls > file_list.txt

# Verify it worked (nothing will print to screen)
cat file_list.txt
```

### The Append Operator (`>>`)
If you want to add new information to the bottom of an existing file *without* erasing the old stuff, use two arrows `>>`.

```bash
echo "This is line 1" > notes.txt
echo "This is line 2" >> notes.txt
```

### The Error Stream (`2>`)
Sometimes programs fail. By default, error messages also print to the screen. However, errors travel on a different "channel" called **Standard Error** (stderr).

If you want to hide errors (or save them to a log file for debugging), you redirect channel #2:

```bash
# Try to list a folder that doesn't exist, and hide the error
ls folder_that_does_not_exist 2> error_log.txt
```

## Part 11 - Process Management: `top`, `ps`, and `kill`

Sometimes a program will crash, freeze, or just take too long to run (common in physics simulations). You need to know how to find these programs and stop them.

### `top`
The `top` command gives you a real-time look at what your computer is doing (like Task Manager in Windows or Activity Monitor on Mac).

```bash
top
```

**What to look for:**
* **PID:** The **Process ID**. This is the unique number assigned to that specific program. You need this number to stop it.
* **COMMAND:** The name of the program (e.g., `python`, `bash`, `chrome`).
* **%CPU:** How much processing power it is eating.

**How to leave:**
Press **`q`** to quit `top`.



### `ps`
If you don't need a live dashboard, `ps` (Process Status) takes a quick snapshot. It is useful for finding the PID of a specific script.

**Common usage:**
```bash
# List processes for the current user (you)
ps -u [your_username]
```

**Finding a specific PID:**
Combine `ps` with `grep` (the filter we learned earlier) to find exactly what you want.
```bash
# Find the PID of my python script
ps -u [your_username] | grep "python"
```

### `kill` - Stopping a Program
Once you know the **PID** of the bad program, you can end it.

**1. The Polite Kill (SIGTERM)**
By default, `kill` asks the program nicely to stop. It allows the program to save data and close files before quitting.
```bash
kill 12345
```
*(Replace 12345 with the actual PID)*

**2. The Force Kill (SIGKILL)**
If the program is frozen solid and ignores the polite request, use the `-9` flag. This cuts the power to the process instantly.

## Part 13 - The Linux Philosophy: Pipes and Filters

One of the most powerful concepts in Linux is the **Unix Philosophy**:
> "Write programs that do one thing and do it well. Write programs to work together."

Instead of building one massive program that does everything (like Microsoft Word), Linux provides hundreds of tiny tools. You combine them using the **Pipe** (`|`).

### The Pipe (`|`)
The pipe key is usually found above the Enter key (Shift + Backslash).
* **What it does:** It takes the **Output** of the command on the left and pushes it as the **Input** into the command on the right.

`Command A` $\rightarrow$ `|` $\rightarrow$ `Command B`

To understand this, we need to meet three new "filter" commands.

### The Filters: `wc`, `head`, and `grep`

1.  **`wc` (Word Count):** Counts lines, words, and characters.
    * `wc -l` counts just the lines.
2.  **`head`:** Shows the first 10 lines of a file.
    * `head -n 5` shows the first 5.
3.  **`grep`:** The search tool. It looks for specific patterns in text.

### Putting it together (The Pipeline)

Let's say you have a folder with hundreds of files, and you want to know how many of them are CSV files.

**Step 1: List everything**
```bash
ls
```
*(Output: A huge list of files...)*

**Step 2: Filter for CSVs**
We pipe the output of `ls` into `grep`.
```bash
ls | grep ".csv"
```
*(Output: Only the files ending in .csv)*

**Step 3: Count them**
We pipe that list into `wc -l` to count the lines.
```bash
ls | grep ".csv" | wc -l
```
*(Output: 42)*

**Why this matters:**
You didn't need to open a window and count icons manually. You combined three "dumb" tools (`ls`, `grep`, `wc`) to answer a specific question. This is how data analysis works on the command line.

## Part 14 - Text Editors


Eventually, you will need to edit code, scripts, or configuration files directly on a server. Since servers often don't have a mouse or a graphical interface, you must use a terminal-based text editor.

There are two main options you will encounter.

### Option A: `nano` (The "Beginner Friendly" Choice)
If you just need to make a quick edit, use `nano`. It behaves like a standard notepad.

1.  **Open a file:** `nano filename.txt`
2.  **Edit:** Just type. You don't need special modes.
3.  **Save:** Press `Ctrl + O` (Write Out), then `Enter`.
4.  **Exit:** Press `Ctrl + X`.

*Note: The menu at the bottom of the screen lists commands. The caret symbol `^` means "Control".*

### Option B: `vi` or `vim` (The "Standard")
`vi` (Visual Editor) is installed on **every** Unix system on earth. If you log into a supercomputer at a national lab, `nano` might not be there, but `vi` will be.
`vim` (Vi IMproved) is the modern version you will likely use.

**The Catch: `vi` is Modal**
`vi` confuses beginners because it has **Modes**. You cannot just open it and start typing.

1.  **Normal Mode (The default):** You use keys to navigate.
    * If you type "exit", you won't write the word "exit"; you might delete lines or jump to the end of the file!
2.  **Insert Mode:** This is for typing text.
3.  **Command Mode:** This is for saving and quitting.

#### The `vi` Survival Guide
Memorize this sequence. It is the "Hello World" of text editing.

1.  **Open the file:**
    ```bash
    vi my_script.sh
    ```
2.  **Enter Insert Mode:**
    Press the **`i`** key.
    *(You can now type text like normal. The bottom of the screen usually says `-- INSERT --`)*.
3.  **Exit Insert Mode:**
    Press the **`Esc`** key.
    *(You are now back in Normal Mode).*
4.  **Save and Quit:**
    Type **`:wq`** and press **Enter**.
    * `:` (Enter command mode)
    * `w` (Write/Save)
    * `q` (Quit)

#### Emergency Exit
If you messed up the file and just want to leave without saving:
1.  Press **`Esc`** (to ensure you are in Normal mode).
2.  Type **`:q!`** and press **Enter**.
    *(The `!` forces it to quit without saving).*

| Editor | Pros | Cons |
| :--- | :--- | :--- |

| **vi / vim** | Available on EVERY computer. Powerful. Fast. | High learning curve. confusing "Modes". |
| **emacs** | Extremely powerful (can run games, email). | Complex key combinations ("Ctrl+Alt+Shift..."). |
